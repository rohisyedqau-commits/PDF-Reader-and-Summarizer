```python
import os
from dotenv import load_dotenv

load_dotenv()
GEMINI_API_KEY = os.getenv("GEMINI_API_KEY")
print("Key loaded:", GEMINI_API_KEY is not None)
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(
    api_key=GEMINI_API_KEY,
    model="gemini-flash-latest",
    temperature=0,
    max_tokens=None,
    timeout=None,
    max_retries=2
)
```

```python
from pypdf import PdfReader

reader = PdfReader("company_profile.pdf")
pdf_text = ""
for page in reader.pages:
    pdf_text += page.extract_text()
print("Total characters extracted:", len(pdf_text))
```

```python
prompt = f"""
You are a professional business analyst.
Summarize the following company profile in simple, clear English.
Include:company pupose, key services, target audience, and unique strenghts.

COMPANY PROFILE TEXT:
{pdf_text}
"""

response = llm.invoke(prompt)
print(response.content)
```
